```python
import pyspark.sql.functions as sf
```

```python
df = (spark.read.format("csv")
                .options(header = True)
                .load("/Volumes/citibike_dev/00_landing_files/citibike_files"))
df = df.withColumn("metadata",
                sf.create_map(
                        sf.lit("pipeline_ID"), sf.lit(1),
                        sf.lit("pipeline_name"), sf.lit(1),
                        sf.lit("pipeline_time"), sf.lit(1)
                ))
# display(df.limit(3))
```

```python
df.write \
  .mode("overwrite") \
  .options(overwriteSchema = True) \
  .saveAsTable("citibike_dev.01_bronze.jc_citibike")
```
